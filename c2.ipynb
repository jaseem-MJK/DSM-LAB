```python
import numpy as np


A = np.array([[3, 1],
              [1, 3]])


U, S, VT = np.linalg.svd(A)

print("Matrix A:")
print(A)

print("\nMatrix U:")
print(U)  

print("\nSingular Values:")
print(S)

print("\nSigma Matrix:")
Sigma = np.zeros((2, 2))
Sigma[0, 0] = S[0]
Sigma[1, 1] = S[1]
print(Sigma)

print("\nMatrix V^T:")
print(VT)


A_reconstructed = U @ Sigma @ VT

print("\nReconstructed Matrix:")
print(A_reconstructed)
```

```python
import matplotlib.pyplot as plt

groups = ["G1", "G2", "G3", "G4", "G5"]
men = [22, 30, 35, 35, 26]
women = [25, 32, 30, 35, 29]

x = [0, 1, 2, 3, 4]

plt.bar(x, men, width=0.4, label="Men", color="blue")
plt.bar([i + 0.4 for i in x], women, width=0.4, label="Women", color="pink")

plt.xticks([i + 0.2 for i in x], groups)
plt.xlabel("Groups")
plt.ylabel("Scores")
plt.title("Scores by Group and Gender")
plt.legend()

plt.show()
```
