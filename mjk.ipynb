```python
n1=int(input("Enter first number: "))
n2=int(input("Enter second number: "))
choice=int(input("Enter a choice:1 addition 2.substraction 3.multiplication 4.division "))
if choice==1:
    addition=n1+n2;
    print(addition)
elif choice==2:
    substraction=n1-n2;
    print(substraction)
elif choice==3:
    multiplication=n1*n2;
    print(multiplication)
elif choice==4:
    division =n1/n2
    print(division)
else:
    print("error")
    
    
```
